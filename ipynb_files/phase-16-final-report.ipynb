## Master Comparison Table — All Models

```python
models = {
    "Spatial CNN\n(EfficientNetB0)"  : SPATIAL_METRICS,
    "Frequency CNN\n(EfficientNetB0)": FREQUENCY_METRICS,
    "Dual-Stream\n(EfficientNetB0)"  : DUAL_EFFNET_METRICS,
    "Dual-Stream\n(Xception)"        : DUAL_XCEPTION_METRICS,
    "FF++ Adapted\n(Xception)"       : FFPP_METRICS,
    "ArtiFact\nZero-Shot"            : ARTIFACT_METRICS,
}

metrics = ["Accuracy", "Precision", "Recall", "F1-Score", "Specificity", "AUC-ROC"]

print("=" * 100)
print("MASTER COMPARISON — ALL MODELS AND DATASETS")
print("=" * 100)
header = f"{'Metric':<14}"
for name in models:
    short = name.replace("\n", " ")[:18]
    header += f" {short:>18}"
print(header)
print("-" * 100)

for m in metrics:
    row = f"{m:<14}"
    vals = [models[n][m] for n in models]
    best = max(vals)
    for v in vals:
        marker = " *" if v == best and v > 0 else "  "
        row += f" {v:>16.4f}{marker}"
    print(row)

print("=" * 100)
print("* = best score for that metric")
```

## Master Comparison Bar Chart

```python
metrics_plot = ["Accuracy", "F1-Score", "Specificity", "AUC-ROC"]
model_names  = ["Spatial\nCNN", "Frequency\nCNN", "Dual\nEffNet",
                "Dual\nXception", "FF++\nAdapted", "ArtiFact\nZero-Shot"]
colors       = ["#4e8cff", "#ff8c4e", "#4ebb78", "#a855f7", "#f59e0b", "#ef4444"]

fig, axes = plt.subplots(1, 4, figsize=(18, 6))
fig.suptitle("Beyond Binary Detection — Master Model Comparison",
             fontsize=14, fontweight="bold")

for ax, metric in zip(axes, metrics_plot):
    vals = [list(models.values())[j][metric] for j in range(len(models))]
    bars = ax.bar(range(len(model_names)), vals, color=colors,
                  edgecolor="black", linewidth=0.5)
    ax.set_xticks(range(len(model_names)))
    ax.set_xticklabels(model_names, fontsize=7.5)
    ax.set_ylim(0, 1.1)
    ax.set_title(metric, fontsize=11, fontweight="bold")
    ax.axhline(y=1.0, color="gray", linestyle="--", linewidth=0.7, alpha=0.5)
    ax.grid(axis="y", alpha=0.3)
    for bar, v in zip(bars, vals):
        if v > 0:
            ax.text(bar.get_x() + bar.get_width()/2, v + 0.01,
                    f"{v:.3f}", ha="center", va="bottom", fontsize=7, rotation=45)

plt.tight_layout()
plt.savefig(os.path.join(RESULT_DIR, "master_comparison.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Master comparison chart saved.")
```

## Answer the Three Research Questions

```python
print("=" * 70)
print("RESEARCH QUESTION ANSWERS")
print("=" * 70)
print()

# Q1 — Accuracy
spatial_auc  = SPATIAL_METRICS["AUC-ROC"]
dual_auc     = DUAL_EFFNET_METRICS["AUC-ROC"]
xception_auc = DUAL_XCEPTION_METRICS["AUC-ROC"]
improvement  = ((dual_auc - spatial_auc) / spatial_auc * 100) if spatial_auc > 0 else 0
print("Q1 — ACCURACY: Can the dual-stream detect hidden AI artifacts?")
print(f"  Spatial CNN AUC     : {spatial_auc:.4f}")
print(f"  Dual-Stream AUC     : {dual_auc:.4f}  (+{improvement:.1f}% over spatial-only)")
print(f"  Xception Dual AUC   : {xception_auc:.4f}")
print(f"  Answer: {'YES — frequency stream adds measurable discriminative signal' if dual_auc > spatial_auc else 'PARTIAL'}")
print()

# Q2 — Transparency
print("Q2 — TRANSPARENCY: Can heatmaps localize manipulation regions?")
print("  Grad-CAM XAI implemented in Phase 11.")
print("  Integrated Gradients saliency maps deployed in Streamlit (Phase 12).")
print("  Answer: YES — spatial attention maps highlight regions driving the prediction.")
print()

# Q3 — Efficiency
print("Q3 — EFFICIENCY: Can research run within 15GB cloud storage?")
print("  All 16 phases ran on Kaggle (free tier, Tesla T4/P100 GPU).")
print("  Stream-and-clean pipeline: only model weights saved permanently.")
print("  Total permanent storage used: model checkpoints only (~200MB per phase).")
print("  Answer: YES — decoupled ingestion architecture confirmed feasible.")
print()

# Q4 — Generalization (bonus — cross-domain result)
artifact_auc = ARTIFACT_METRICS["AUC-ROC"]
print("BONUS — GENERALIZATION: Does the model transfer to unseen generators?")
print(f"  ArtiFact zero-shot AUC : {artifact_auc:.4f}  (25 generators, no training)")
gen_verdict = (
    "YES — strong cross-generator generalization" if artifact_auc >= 0.75
    else "PARTIAL — generalizes but generator-sensitive" if artifact_auc >= 0.65
    else "NEEDS WORK — insufficient generalization"
)
print(f"  Answer: {gen_verdict}")
print()
print("=" * 70)
```
